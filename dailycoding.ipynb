```python
def result(lis, k):
  n = len(lis)
  for i in range(n):
    for j in range(i + 1, n):
      if lis[i] + lis[j] == k:
        return True
  return False

a = [10, 15, 3, 7]
b = 17
print(result(a, b))  # Expected output: True
```

```python
def product(arr):
  new = []
  for i in range(0, len(arr)): # to iterate the array
    curr = i
    sum = 1
    for j in range(0, len(arr)): # to find the product
      if j != curr:
        sum = sum*arr[j]
    new += [sum]
  return new

a = [1, 2, 3, 4, 5]
print(product(a)) # Expected output: [120, 60, 40, 30, 24]
b= [3, 2, 1]
print(product(b))  # Expected output: [2, 3, 6]
```

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def serialize(node):
    if node is None:
        return '#'
    return node.val + ' ' + serialize(node.left) + ' ' + serialize(node.right)

def deserialize(data):
    values = data.split()
    index = [0]

    def deserialize_node():
        if values[index[0]] == '#':
            index[0] += 1
            return None

        node = Node(values[index[0]])
        index[0] += 1
        node.left = deserialize_node()
        node.right = deserialize_node()
        return node

    return deserialize_node()

# Build tree
node = Node('root', Node('left', Node('left.left')), Node('right'))

# Serialize
serialized_str = serialize(node)
print("Serialized tree:", serialized_str)

# Deserialize
new_tree = deserialize(serialized_str)

# Check value
print("Left.Left node value:", new_tree.left.left.val)

# Optional: Assertion check
assert new_tree.left.left.val == 'left.left'
```

```python
def prefix(lis, qstr):
  new = []
  for word in lis:
    if word.startswith(qstr):
      new.append(word)
  return new

lis = ['dog', 'deer', 'deal']
qstr = 'de'
print(prefix(lis, qstr))  # Expected output: ['deer', 'deal']
```
